 Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

```python
x1 = int (input('Enter the value of x1: '))
y1 = int (input('Enter the value of y1: '))
x2 = int (input('Enter the value of x2: '))
y2 = int (input('Enter the value of y2: '))

distance = ((y2-y1)**2 + (x2-x1)**2)**0.5

print('The distance between the co ordinates is: ' , distance)
```

Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

```python
P = float(input('Enter the value of Principle: '))
R = float(input('Enter the rate of the principle: '))
T = float(input('Enter the time period for the principle: '))

SI = P * T * R

print ('The interest rate of the principle for give period of time is: ', SI)
```

Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

```python
heads = int(input('Enter the number of heads: '))
legs = int(input('Enter the numbr of legs: '))
dogs = (legs - 2 * heads) // 2
chickens = heads - dogs
print ('The number of dogs are: ', dogs)
print ('The number of chickens are: ', chickens)

```

Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

```python
a1 = int(input('Enter the first term of the series: '))
a2 = int(input('Enter the second term of the series: '))
n = int(input("Enter the term position (n) to find: "))

d = a2 - a1

an = a1 + (n-1) * d

print('The Nth term of the series is: ', an)
```

Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

```python
a = float(input('Enter the first number numerator: '))
b = float(input('Enter the first number denominator: '))
c = float(input('Enter the second number numerator: '))
d = float(input('Enter the second number denominator: '))

i = a / b
j = c / d

result = i + j

print('The sum of the fractions is: ', result)
```

Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.

Input:

Dimensions of the milk tank
H = 20cm, L = 20cm, B = 20cm

Dimensions of the glass
h = 3cm, r = 1cm

```python
import math

a = float(input('Enter the height of the milk tank: '))
b = float(input('Enter the width of the milk tank: '))
c = float(input('Enter the breadth of the milk tank: '))

i = float(input('Enter the height of the glass: '))
j = float(input('Enter the radius of the glass: '))

tank_volume = a * b * c
glass_volume = math.pi * (j**2) * i

result = float(tank_volume // glass_volume)

print('The number of glasses that can be obtained are: ', result)
```
