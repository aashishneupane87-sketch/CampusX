Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

```python
a = int (input('Enter 1st Number: '))
b = int (input('Enter 2nd Number: '))
a, b = b, a
print ("the swapped number is:", a)
print ("the second swapped number is: ", b)
```

 Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

```python
x1 = int (input('Enter the value of x1: '))
y1 = int (input('Enter the value of y1: '))
x2 = int (input('Enter the value of x2: '))
y2 = int (input('Enter the value of y2: '))

distance = ((y2-y1)**2 + (x2-x1)**2)**0.5

print('The distance between the co ordinates is: ' , distance)
```

Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

```python
P = float(input('Enter the value of Principle: '))
R = float(input('Enter the rate of the principle: '))
T = float(input('Enter the time period for the principle: '))

SI = P * T * R

print ('The interest rate of the principle for give period of time is: ', SI)
```
